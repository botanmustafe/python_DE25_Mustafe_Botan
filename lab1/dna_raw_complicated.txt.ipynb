```python
# How to open the file safely and how to read every line
with open("dna_raw_complicated.txt") as f:
    data = f.readlines()


sequences = {}          # Dictionary to store all DNA sequences
sequence_name = None    # Keeps track of current sequence name
sequence_dna = ""       # Keeps track of DNA lines for the current sequence


# Loop through every line in the file
for line in data:
    line = line.strip()    # Remove extra spaces

    if line.startswith(">"): # Check if the line begins with > meaning a new sequence starts here.
              # If we already have a sequence, save it to the dictionary       
            if sequence_name:
              sequences[sequence_name] = sequence_dna
                 # Set the new sequence name (remove ">"), reset DNA string for the new sequence
              sequence_name = line.strip()[1:]
              sequence_dna = ""
        else:
              sequence_dna += line # Add the DNA line to the sequence

              # After the loop, save the last sequence
              if sequence_name:
                    sequences[sequence_name] = sequence_dna


              





        
```
