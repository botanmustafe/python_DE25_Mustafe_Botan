# Clean the quotes file

```python
# we will use regular expression package to clean string contents
import re
```

```python
# create an empty list for saving quotes
quotes = []

# initiate a quote number
i = 1
```

```python
# open the file 
# comparing the use of absolute and relative paths

# print out all the contents at once
raw_file_path = "quotes.txt"

# loop through lines in the file contents
with open(raw_file_path, "r") as f_read:
    quotes = f_read.read()
    
    print(f_read) # what you can see from just opening the file
    print(quotes) # what you can see from applying the read() method on the same file
```

```python
import re
i = 1

# open two files (one with r mode & another with w mode)
# loop through lines in the file contents
with open(raw_file_path, "r") as f_read, open("quotes_cleaned.txt", "w") as f_write:
    for quote in f_read:
        # data cleaning
        quote = quote.strip(" \n") # replace the quote with modified quote
        quote = re.sub(" +", " ", quote) # replace multiple spaces with one space
        # print each line
        print(quote)

        if quote != "":
            f_write.write(f"{i}. {quote}\n")
            i += 1
        
```
